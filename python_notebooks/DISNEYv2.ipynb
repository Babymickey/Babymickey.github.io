```python
import pandas as pd
disney_df = pd.read_csv('https://query.data.world/s/q6d4wfxrzhpcxwurhr5ly6hv7vb7ti?dws=00000')
disney_df
```

```python
disney_df['release_date'] = pd.to_datetime(disney_df['release_date'])
disney_df['total_gross'] = (
    disney_df['total_gross']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
import altair as alt

genre_color_map = {
    'Action': '#4fc2ff',
    'Adventure': '#1670ad',
    'Black Comedy': '#a7ff58',
    'Comedy': '#33a02c',
    'Concert/Performance': '#efa4a4',
    'Documentary': '#e31a1c',
    'Drama': '#fdbf6f',
    'Horror': '#ff7f00',
    'Musical': '#d889ff',
    'Romantic Comedy': '#6a3d9a',
    'Thriller/Suspense': '#ffff57',
    'Western': '#b15928'
}

genres_list = list(genre_color_map.keys())
colors_list = list(genre_color_map.values())

genre_dropdown = alt.binding_select(
    options=[None] + genres_list, 
    labels=['All Genres'] + genres_list, 
    name="Select Genre: "
)
genre_select = alt.selection_point(fields=['genre'], bind=genre_dropdown)

main_color_chart = alt.Chart(disney_df).mark_circle(size=100).encode(
    x=alt.X('release_date:T', title='Release Year'),
    y=alt.Y('total_gross:Q', title='Total Gross (USD)', axis=alt.Axis(format='$.2s')),
    color=alt.Color('genre:N', 
        title='Genre',
        scale=alt.Scale(
            domain=genres_list,  
            range=colors_list    
        )
    ),
    tooltip=['movie_title', 'release_date', 'total_gross', 'genre']
).add_params(
    genre_select
).transform_filter(
    genre_select
).properties(
    width=1000,
    height=900,
    title="The Evolution of Disney Box Office Success"
)

main_color_chart
```

```python
import altair as alt

disney_cleaned = disney_df.dropna(subset=['genre'])

genre_color_map = {
    'Action': '#4fc2ff',
    'Adventure': '#1670ad',
    'Black Comedy': '#a7ff58',
    'Comedy': '#33a02c',
    'Concert/Performance': '#efa4a4',
    'Documentary': '#e31a1c',
    'Drama': '#fdbf6f',
    'Horror': '#ff7f00',
    'Musical': '#d889ff',
    'Romantic Comedy': '#6a3d9a',
    'Thriller/Suspense': '#ffff57',
    'Western': '#b15928'
}

genres_list = list(genre_color_map.keys())
colors_list = list(genre_color_map.values())

genre_color_chart = alt.Chart(disney_cleaned).mark_bar().encode(
    x=alt.X('average(total_gross):Q', title='Avg Gross', axis=alt.Axis(format='$.2s')),
    y=alt.Y('genre:N', sort='-x', title='Genre'),
    color=alt.Color('genre:N', 
        scale=alt.Scale(
            domain=genres_list, 
            range=colors_list
        ),
        legend=None 
    )
).properties(
    width=1000, 
    height=900, 
    title="Which Genres Make the Most?"
)

genre_color_chart
```

```python
import altair as alt

renaissance_df = disney_df[(disney_df['release_date'] >= '1989-01-01') & (disney_df['release_date'] <= '1999-12-31')]
renaissance_cleaned = renaissance_df.dropna(subset=['genre']) 

top_30_renaissance = renaissance_cleaned.nlargest(30, 'total_gross')

genre_color_map = {
    'Action': '#4fc2ff',
    'Adventure': '#1670ad',
    'Black Comedy': '#a7ff58',
    'Comedy': '#33a02c',
    'Concert/Performance': '#efa4a4',
    'Documentary': '#e31a1c',
    'Drama': '#fdbf6f',
    'Horror': '#ff7f00',
    'Musical': '#d889ff',
    'Romantic Comedy': '#6a3d9a',
    'Thriller/Suspense': '#ffff57',
    'Western': '#b15928'
}

genres_list = list(genre_color_map.keys())
colors_list = list(genre_color_map.values())

renaissance_color_chart = alt.Chart(top_30_renaissance).mark_bar().encode(
    x=alt.X('total_gross:Q', title='Total Gross (USD)', axis=alt.Axis(format='$.2s')),
    y=alt.Y('movie_title:N', sort='-x', title=None),
       
    color=alt.Color('genre:N', 
        title="Genre",
        scale=alt.Scale(
            domain=genres_list, 
            range=colors_list
        )
    ),
     
    tooltip=['movie_title', 'genre', 'release_date', 'total_gross']
).properties(
    width=1000,
    height=900,
    title="Top 30 Hits of the Disney Renaissance (1989-1999)"
)

renaissance_color_chart
```

```python
boxoffice = pd.read_csv("https://raw.githubusercontent.com/kalilurrahman/BoxOfficeData/main/boxofficemojoustop1000.tsv", sep='\t')
boxoffice
```

```python
import pandas as pd
import altair as alt

boxoffice = boxoffice.copy()

boxoffice['Lifetime Gross'] = pd.to_numeric(boxoffice['Lifetime Gross'], errors='coerce')
boxoffice['Year'] = pd.to_numeric(boxoffice['Year'], errors='coerce')

renaissance_df = boxoffice[(boxoffice['Year'] >= 1989) & (boxoffice['Year'] <= 2022)]

top_25_renaissance = renaissance_df.nlargest(25, 'Lifetime Gross')

boxoffice_chart = alt.Chart(top_25_renaissance).mark_bar().encode(
    x=alt.X('Lifetime Gross:Q', title='Lifetime Gross (USD)', axis=alt.Axis(format='$.2s')),
    y=alt.Y('Movie:N', sort='-x', title=None), # Changed from 'Title' to 'Movie'
    color=alt.Color('Lifetime Gross:Q', scale=alt.Scale(scheme='blues'), legend=None),
    tooltip=['Movie', 'Year', 'Lifetime Gross']
).properties(
    width=800,
    height=600,
    title="Top 25 Box Office Hits (1989-1999)"
)

boxoffice_chart
```

```python
import pandas as pd
import altair as alt

boxoffice = boxoffice.copy()
boxoffice['Lifetime Gross'] = pd.to_numeric(boxoffice['Lifetime Gross'], errors='coerce')

movie_to_genre = {
    # Star Wars (Action - Light Blue)
    'Star Wars: Episode VII - The Force Awakens': 'Action',
    'Star Wars: Episode VIII - The Last Jedi': 'Action',
    'Star Wars: Episode IX - The Rise of Skywalker': 'Action',
    'Star Wars: Episode I - The Phantom Menace': 'Action',
    'Rogue One: A Star Wars Story': 'Action',
    
    # Marvel (Action - Light Blue)
    'Avengers: Endgame': 'Action',
    'Spider-Man: No Way Home': 'Action',
    'Black Panther': 'Action',
    'Avengers: Infinity War': 'Action',
    'The Avengers': 'Action',
    'Avengers: Age of Ultron': 'Action',
    'Captain Marvel': 'Action',

    # Animation/Other (Musical - Purple, Adventure - Dark Blue)
    'Incredibles 2': 'Adventure',
    'The Lion King': 'Adventure',
    'Beauty and the Beast': 'Musical',
    'Finding Dory': 'Adventure',
    'Frozen II': 'Musical',
    'Toy Story 4': 'Adventure'
}

boxoffice['genre'] = boxoffice['Movie'].map(movie_to_genre)

boxoffice['genre'] = boxoffice['genre'].fillna('Not Disney')

genre_color_map = {
    'Action': '#4fc2ff', 
    'Adventure': '#1670ad', 
    'Black Comedy': '#a7ff58',
    'Comedy': '#33a02c', 
    'Concert/Performance': '#efa4a4', 
    'Documentary': '#e31a1c',
    'Drama': '#fdbf6f', 
    'Horror': '#ff7f00', 
    'Musical': '#d889ff',
    'Romantic Comedy': '#6a3d9a', 
    'Thriller/Suspense': '#ffff57', 
    'Western': '#b15928',
    'Not Disney': '#cccccc' 
}

genres_list = list(genre_color_map.keys())
colors_list = list(genre_color_map.values())

top_25_data = boxoffice.nlargest(25, 'Lifetime Gross')

boxoffice_color_chart = alt.Chart(top_25_data).mark_bar().encode(
    x=alt.X('Lifetime Gross:Q', title='Lifetime Gross (USD)', axis=alt.Axis(format='$.2s')),
    y=alt.Y('Movie:N', sort='-x', title=None),
    color=alt.Color('genre:N', 
        title="Genre Key", 
        scale=alt.Scale(domain=genres_list, range=colors_list),
        legend=alt.Legend(orient='right', offset=20)
    ),
    tooltip=['Movie', 'Year', 'Lifetime Gross', 'genre']
).properties(
    width=1000,
    height=900,
    title="Top 25 Box Office Hits: Disney Highlighted by Genre"
)

boxoffice_color_chart
```
